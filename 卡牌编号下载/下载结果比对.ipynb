```python
import requests
from pathlib import Path

for now in range(1000, 2056):
    url = f"https://nie.res.netease.com/nie/hlbt/zfj/cards/{now}.png?v=1"
    try:
        res = requests.get(url, verify=False)
        if res.status_code != 200:
            print(f"Failed to download {now}.png: Status code {res.status_code}")
            continue
        else:
            output_dir = Path("小程序下载结果")
            output_dir.mkdir(exist_ok=True)
            with open(output_dir / f"{now}.png", "wb") as f:
                f.write(res.content)
            print(f"Downloaded {now}.png")
    except Exception as e:
        print(f"Failed to download {now}.png: {e}")
```

```python
import os
import hashlib
from pathlib import Path

def get_image_files(folder):
    return {f.name: f for f in Path(folder).glob("*") if f.is_file()}

def get_file_hash(filepath, algo="md5"):
    h = hashlib.new(algo)
    with open(filepath, "rb") as f:
        while chunk := f.read(8192):
            h.update(chunk)
    return h.hexdigest()

# 文件夹路径
folder1 = "result"
folder2 = "小程序下载结果"

# 获取文件映射：文件名 -> Path 对象
files1 = get_image_files(folder1)
files2 = get_image_files(folder2)

# 所有文件名集合
all_filenames = set(files1.keys()).union(files2.keys())

only_in_1 = []
only_in_2 = []
different_content = []

for filename in all_filenames:
    f1 = files1.get(filename)
    f2 = files2.get(filename)

    if f1 and not f2:
        only_in_1.append(filename)
    elif f2 and not f1:
        only_in_2.append(filename)
    else:
        # 两边都有这个文件名，比较内容
        hash1 = get_file_hash(f1)
        hash2 = get_file_hash(f2)
        if hash1 != hash2:
            different_content.append(filename)

# 打印结果
print("【只在 result 中的文件】:")
for f in only_in_1:
    print(f)

print("\n【只在 小程序下载结果 中的文件】:")
for f in only_in_2:
    print(f)

print("\n【文件名相同但内容不同】:")
for f in different_content:
    print(f)
```
